```python
import pandas as pd

# 讀取所有工作表
excel_path = "/root/.cache/kagglehub/datasets/truongdai/tableau-sample-superstore/versions/1/Sample - Superstore.xls"
sheets = pd.read_excel(excel_path, sheet_name=None)

# 把每個工作表轉成不同的變數
Orders = sheets[list(sheets.keys())[0]]
People = sheets[list(sheets.keys())[1]]
Return = sheets[list(sheets.keys())[2]]

# 檢查一下每個 df 的前幾列
print("Orders preview:")
print(Orders.head())
print("People preview:")
print(People.head())
print("Return preview:")
print(Return.head())
```

# 篩選 2021 年

```python
# 確保 Order Date 是 datetime 格式
Orders['Order Date'] = pd.to_datetime(Orders['Order Date'])

# 篩選 2021 年的資料
Orders_2021 = Orders[Orders['Order Date'].dt.year == 2021]

# 顯示結果預覽
print("2021 年的訂單筆數:", len(Orders_2021))
print(Orders_2021.head())
```

```python
import pandas as pd

# 確保日期格式正確
Orders_2021["Order Date"] = pd.to_datetime(Orders_2021["Order Date"])

# 對 Return 做 Order ID 去重（避免重複值影響 join 結果）
Return_unique = Return.drop_duplicates(subset=["Order ID"])

# 進行左連接
Orders_with_returns = pd.merge(
    Orders_2021,
    Return_unique,
    how="left",
    on="Order ID"
)

# 顯示範例
print("含退貨資訊的 Orders：")
print(Orders_with_returns.head())
```

```python
len(Orders_with_returns)
```

```python
# 篩選出未退貨的資料
non_returned_orders = Orders_with_returns[Orders_with_returns['Returned'].isnull()]

# 計算未退貨資料的總 Sales
total_sales_non_returned = non_returned_orders['Sales'].sum()

# 顯示結果
print(f"Orders_with_returns 中扣除退貨後的總 Sales 金額為: {total_sales_non_returned:.2f}")
```

```python
# prompt: Orders_with_returns 中 12 月份扣除退貨後的總 Sales 金額為

import pandas as pd
# 確保日期格式正確
Orders_with_returns["Order Date"] = pd.to_datetime(Orders_with_returns["Order Date"])

# 篩選出 12 月份的資料
december_orders = Orders_with_returns[Orders_with_returns['Order Date'].dt.month == 12]

# 篩選出 12 月份且未退貨的資料
december_non_returned = december_orders[december_orders['Returned'].isnull()]

# 計算 12 月份未退貨資料的總 Sales
total_sales_december_non_returned = december_non_returned['Sales'].sum()

# 顯示結果
print(f"Orders_with_returns 中 12 月份扣除退貨後的總 Sales 金額為: {total_sales_december_non_returned:.2f}")
```
